# Generate Lemonite Summary Report — Wang GBM

This notebook generates a `Lemonite_Summary_Report.html` for the Wang GBM dataset by running the same `generate_summary_report.py` script used by the nextflow pipeline.

The script expects a `LemonTree/` subdirectory under `output_dir`. Because this dataset's results live *inside* a directory named `LemonTree/`, we create a temporary `LemonTree` symlink inside the data directory pointing to itself so the script's path resolution works correctly. The symlink is removed after the report is generated.

**Output:** `Lemonite_Summary_Report.html` saved to the data directory.

```python
import os
import subprocess
import sys

# ── Paths ──────────────────────────────────────────────────────────────────────
SCRIPT    = '/home/borisvdm/repo/LemonIte/nextflow/scripts/generate_summary_report.py'

# The run's result directory (contains Preprocessing/, Lemon_out/, Networks/, ...)
DATA_DIR  = '/home/borisvdm/Documents/PhD/thesis_Mirte/Wang2021/results/LemonTree/noProteomics_percentile2_divide_by_sum'

# output_dir = DATA_DIR; run_id='.' → script looks in DATA_DIR/LemonTree/{...}
# We satisfy this by symlinking DATA_DIR/LemonTree -> DATA_DIR
OUTPUT_DIR = DATA_DIR
INPUT_DIR  = '/home/borisvdm/Documents/PhD/thesis_Mirte/Wang2021'

RUN_ID    = '.'
ORGANISM  = 'human'

# Regulator prefixes must match file stems in Lemon_out/ (e.g. Lovering.percentile2.txt)
REGULATOR_TYPES = 'Lovering:Lovering_TF_list.txt,Metabolite:Metabolomics.txt,Lipid:Lipidomics.txt'

# name_mapping.tsv (optional — for restoring original feature names)
NAME_MAPPING = os.path.join(DATA_DIR, 'Preprocessing', 'name_mapping.tsv')

print(f'Data dir    : {DATA_DIR}')
print(f'Script      : {SCRIPT}')
print(f'name_mapping: {NAME_MAPPING} (exists={os.path.exists(NAME_MAPPING)})')
```

```python
# ── Set up directory structure ─────────────────────────────────────────────────

# 1. DATA_DIR/LemonTree -> DATA_DIR  (so script finds DATA_DIR/LemonTree/Preprocessing etc.)
lemontree_link = os.path.join(DATA_DIR, 'LemonTree')
if not os.path.exists(lemontree_link):
    os.symlink(DATA_DIR, lemontree_link)
    print(f'Created symlink: {lemontree_link} -> {DATA_DIR}')
else:
    print(f'LemonTree symlink/dir already exists: {lemontree_link}')

# 2. Module_Overview/Module_Overview.csv -> existing Module_Overview_Complete.csv
module_overview_dir = os.path.join(DATA_DIR, 'Module_Overview')
os.makedirs(module_overview_dir, exist_ok=True)

target_csv = os.path.join(DATA_DIR, 'Module_Overview_Complete.csv')
link_csv   = os.path.join(module_overview_dir, 'Module_Overview.csv')

if os.path.exists(target_csv):
    if not os.path.exists(link_csv):
        os.symlink(target_csv, link_csv)
        print(f'Created symlink: {link_csv} -> {target_csv}')
    else:
        print(f'Module_Overview.csv already exists: {link_csv}')
else:
    print(f'WARNING: source CSV not found: {target_csv}')

print('Directory setup complete.')
```

```python
# ── Run generate_summary_report.py ────────────────────────────────────────────
cmd = [
    sys.executable, SCRIPT,
    '--input_dir',       INPUT_DIR,
    '--output_dir',      OUTPUT_DIR,
    '--run_id',          RUN_ID,
    '--regulator_types', REGULATOR_TYPES,
    '--organism',        ORGANISM,
]

if os.path.exists(NAME_MAPPING):
    cmd += ['--name_mapping', NAME_MAPPING]

print('Running:', ' '.join(cmd))
print('-' * 80)

result = subprocess.run(cmd, capture_output=False, text=True)

print('-' * 80)
print(f'Exit code: {result.returncode}')
```

```python
# ── Clean up temporary LemonTree symlink ──────────────────────────────────────
if os.path.islink(lemontree_link):
    os.unlink(lemontree_link)
    print(f'Removed symlink: {lemontree_link}')

# ── Verify output ─────────────────────────────────────────────────────────────
output_html = os.path.join(OUTPUT_DIR, 'Lemonite_Summary_Report.html')

if os.path.exists(output_html):
    size_mb = os.path.getsize(output_html) / 1e6
    print(f'Report generated successfully!')
    print(f'  Path : {output_html}')
    print(f'  Size : {size_mb:.1f} MB')
else:
    print(f'ERROR: Report not found at {output_html}')
```
