# Generate Lemonite Summary Report — Lloyd-Price IBD

This notebook generates a `Lemonite_Summary_Report.html` for the Lloyd-Price IBD dataset by running the same `generate_summary_report.py` script used by the nextflow pipeline.

**Output:** `Lemonite_Summary_Report.html` saved to the results directory.

```python
import os
import subprocess
import sys

# ── Paths ──────────────────────────────────────────────────────────────────────
SCRIPT      = '/home/borisvdm/repo/LemonIte/nextflow/scripts/generate_summary_report.py'
LEMON_DIR   = '/home/borisvdm/Documents/PhD/Lemonite/Lloyd-Price_IBD/results/LemonTree'
OUTPUT_DIR  = '/home/borisvdm/Documents/PhD/Lemonite/Lloyd-Price_IBD/results'
INPUT_DIR   = '/home/borisvdm/Documents/PhD/Lemonite/Lloyd-Price_IBD'

# run_id='.' → script looks in OUTPUT_DIR/LemonTree/{Preprocessing,Lemon_out,...}
RUN_ID      = '.'
ORGANISM    = 'human'

# Regulator prefixes must match the file stems in Lemon_out/ (e.g. Lovering.percentile2.txt)
REGULATOR_TYPES = 'Lovering:Lovering_TF_list.txt,Metabolite:Metabolomics.txt'

# name_mapping.tsv (optional — for restoring original feature names)
NAME_MAPPING = os.path.join(LEMON_DIR, 'Preprocessing', 'name_mapping.tsv')

print(f'LemonTree dir : {LEMON_DIR}')
print(f'Output dir    : {OUTPUT_DIR}')
print(f'Script        : {SCRIPT}')
print(f'name_mapping  : {NAME_MAPPING} (exists={os.path.exists(NAME_MAPPING)})')
```

```python
# ── Set up Module_Overview directory ──────────────────────────────────────────
# The script looks for OUTPUT_DIR/LemonTree/Module_Overview/Module_Overview.csv.
# The existing file has a different name; create a symlink so the script finds it.

module_overview_dir = os.path.join(LEMON_DIR, 'Module_Overview')
os.makedirs(module_overview_dir, exist_ok=True)

target_csv = os.path.join(LEMON_DIR, 'Module_Overview_Complete_Expression_UC_vs_nonIBD.csv')
link_csv   = os.path.join(module_overview_dir, 'Module_Overview.csv')

if os.path.exists(target_csv):
    if not os.path.exists(link_csv):
        os.symlink(target_csv, link_csv)
        print(f'Created symlink: {link_csv} -> {target_csv}')
    else:
        print(f'Module_Overview.csv already exists: {link_csv}')
else:
    print(f'WARNING: source CSV not found: {target_csv}')

# Also symlink Module_coherence_scores.txt into Networks/ if not already there
coherence_src  = os.path.join(LEMON_DIR, 'Module_coherence_scores.txt')
coherence_dest = os.path.join(LEMON_DIR, 'Networks', 'Module_coherence_scores.txt')
if os.path.exists(coherence_src) and not os.path.exists(coherence_dest):
    os.symlink(coherence_src, coherence_dest)
    print(f'Created symlink: {coherence_dest} -> {coherence_src}')

print('Directory setup complete.')
```

```python
# ── Run generate_summary_report.py ────────────────────────────────────────────
cmd = [
    sys.executable, SCRIPT,
    '--input_dir',       INPUT_DIR,
    '--output_dir',      OUTPUT_DIR,
    '--run_id',          RUN_ID,
    '--regulator_types', REGULATOR_TYPES,
    '--organism',        ORGANISM,
]

if os.path.exists(NAME_MAPPING):
    cmd += ['--name_mapping', NAME_MAPPING]

print('Running:', ' '.join(cmd))
print('-' * 80)

result = subprocess.run(cmd, capture_output=False, text=True)

print('-' * 80)
print(f'Exit code: {result.returncode}')
```

```python
# ── Verify output ─────────────────────────────────────────────────────────────
output_html = os.path.join(OUTPUT_DIR, 'Lemonite_Summary_Report.html')

if os.path.exists(output_html):
    size_mb = os.path.getsize(output_html) / 1e6
    print(f'Report generated successfully!')
    print(f'  Path : {output_html}')
    print(f'  Size : {size_mb:.1f} MB')
else:
    print(f'ERROR: Report not found at {output_html}')
```
